```python
from pyspark.sql import functions as F

caminho_cetic = (
    "/Volumes/workspace/mvp_bronze/raw_files/cetic_2025/"
    "tic_domicilios_2025_individuos_base_de_microdados_v1.0.csv"
)

df_cetic = (
    spark.read
    .option("header", True)
    .option("sep", ";")
    .option("inferSchema", False)
    .csv(caminho_cetic)
)
```

```python
df_cetic_bronze = (
    df_cetic
    .withColumn("_data_ingestao", F.current_timestamp())
    .withColumn("_fonte", F.lit("CETIC.br - TIC Domicilios 2025"))
    .withColumn("_ano_referencia", F.lit(2025))
    .withColumn(
        "_arquivo_origem",
        F.lit("tic_domicilios_2025_individuos_base_de_microdados_v1.0.csv")
    )
)
```

```python
(
    df_cetic_bronze.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(
        "workspace.mvp_bronze.cetic_individuos_2025"
    )
)
```
